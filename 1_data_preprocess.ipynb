# TPC

```python
rna = anndata.read_h5ad("../TPC.h5ad")
meta = pd.read_csv('../TPC_meta.csv', index_col=0)

sc.pp.filter_cells(rna, min_genes=1)
sc.pp.filter_genes(rna, min_cells=200)

rna.obs = meta

rna.raw = rna
# convert the adata object into a dataframe which has rownames as cell ID, 
# colnames as gene names
df = pd.DataFrame(data=rna.raw.X, 
                  index=rna.raw.obs_names, 
                  columns=rna.raw.var_names)
df.head()
```

```python
gene_df = rna.var
gene_filtered = list(gene_df[gene_df['n_cells']>0].index)
adata_filtered = df[gene_filtered]
adata_filtered.head()
```

```python
adata_filtered = adata_filtered[0:-1]
merged_features = pd.merge(pd.DataFrame(rna.obs['BMI']), adata_filtered,
                           left_index=True, right_index=True)
merged_features = pd.merge(merged_features,pd.DataFrame(rna.obs['age']), 
                           left_index=True, right_index=True)
merged_features.to_csv('../TPC_data.csv')
```
